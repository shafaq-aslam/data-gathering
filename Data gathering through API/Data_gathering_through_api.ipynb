# Data Gathering Through API 

What is APIs?
- APIs helps to communicate to softwares or APIs nothing but Data pipelines.

```python
import pandas as pd
import requests
```

```python
response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=1a4fe1332d115268754c6a603ce1e673&language=en-US&page=1')
```

```python
response.json()
```

```python
response.json()['results']
```

```python
# to convert results in DataFrame
pd.DataFrame(response.json()['results']).head()
```

```python
# to fatch desired columns for the dataframe
df = pd.DataFrame(response.json()['results'])[['id', 'title', 'release_date', 'popularity', 'vote_average', 'vote_count', 'overview']]
```

```python
df
```

```python
## to fatch data from all 428 pages to create a complete dataset
```

```python
for i in range(1, 429):
    response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=1a4fe1332d115268754c6a603ce1e673&language=en-US&page={}'.format(i))
    temp_df = pd.DataFrame(response.json()['results'])[['id', 'title', 'release_date', 'popularity', 'vote_average', 'vote_count', 'overview']]
    df = pd.concat([df, temp_df], ignore_index=True) # used ignore_index = True to avoid repeatation of index
```

```python
df
```

```python
df.to_csv('top_rated_movies.csv')
```
