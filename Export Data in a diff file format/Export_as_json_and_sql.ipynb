# Export Data as JSON and SQL

```python
import pandas as pd
```

```python
df = pd.read_csv('deliveries.csv')
df.head()
```

## to_json

```python
temp_df = df.groupby(['batting_team', 'batsman'])['batsman_runs'].sum().unstack()
temp_df
```

```python
temp_df.to_json('ipl.json')
```

```python
import pymysql
from sqlalchemy import create_engine
```

```python
engine = create_engine("mysql+pymysql://root:@localhost/match_data")
# mysql+pymysql://{root}:{password}@{url}/{database_name}
df.to_sql('match_delivery', con = engine, if_exists = 'append')
```

```python
temp_df = df.groupby('batsman')['batsman_runs'].sum().reset_index()
temp_df.head()
```

```python
temp_df.to_sql('batsman_runs', con = engine, if_exists = 'append')
```

```python
temp_df2 = df.query('batsman_runs == 6').pivot_table(index='over',columns='ball',values='batsman_runs',aggfunc='count')
temp_df2
```

```python
temp_df2.to_sql('six_heatmap', con = engine, if_exists = 'append')
```
