```python
import pandas as pd
```

```python
df = pd.read_csv('deliveries.csv')
df.head()
```

```python
temp_df1 = df.query('batsman_runs == 6').pivot_table(index = 'over', columns = 'ball', values = 'batsman_runs', aggfunc = 'count')
temp_df1
```

```python
temp_df1.to_html('sixer_heatmap.html')
```
