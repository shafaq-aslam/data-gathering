## 3. Opening a CSV file from an URL

```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

## 4. Parameters

### Sep Parameter

```python
# rather than finding coma, separate by \t (tab)
pd.read_csv('movie_titles_metadata.tsv', sep = '\t')
```

### names Parameter

```python
# if you have no column names or column names are not defined.
pd.read_csv('movie_titles_metadata.tsv', sep = '\t', names = ['s_no', 'movie_name', 'release_year', 'rating', 'votes', 'genres'])
```

### Index_col Parameter

```python
pd.read_csv('movie_titles_metadata.tsv', sep = '\t', names = ['s_no', 'movie_name', 'release_year', 'rating', 'votes', 'genres'], index_col = 's_no')
```

### Header Parameter 

```python
# important
pd.read_csv('test.csv') # before header
```

```python
pd.read_csv('test.csv', header = 1) # after header
```

### use_cols Parameter

```python
pd.read_csv('aug_train.csv')
```

```python
# if you do not need all the columns in the data, just the specific ones..
pd.read_csv('aug_train.csv', usecols = ['enrollee_id', 'gender', 'education_level', 'major_discipline'])
```

### squeeze Parameters

```python
# when you don't need all columns but one.
# it will give you pandas series instead of dataframe

"""
sqeeze Parameter has deprecated instead of using "squeeze = True",
use squeeze() function on the dataframe.
"""
pd.read_csv('aug_train.csv', usecols= ['major_discipline']).squeeze() 
```

### skiprows/nrow Parameter

```python
pd.read_csv('aug_train.csv')
```

```python
pd.read_csv('aug_train.csv', skiprows = [0, 1])
```

```python
pd.read_csv('aug_train.csv', skiprows = lambda x : x%3 == 0)
```

```python
pd.read_csv('aug_train.csv', nrows = 100)
```

### Encoding Parameter

```python
pd.read_csv('zomato.csv', encoding = 'latin-1')
```

### On bad lines

```python
pd.read_csv('zomato.csv', encoding="latin-1" , on_bad_lines='skip') # will skip problematic lines
```
